### 1. Load and Explore the Food Delivery Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


df = pd.read_csv('/Food_Delivery_Times.csv')


print(df.head())
print("\nDataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nSummary statistics:\n", df.describe())
```

### 2. Prepare Features and Target Variable
We will dynamically select the target column (typically 'Delivery_Time' or similar target present in the dataset) and encode any categorical variables if they exist.

```python

target = 'Delivery_Time_min'
print(f"Selected target variable: {target}")


features = [c for c in df.columns if c not in [target, 'Order_ID']]


data = df.dropna()


X = pd.get_dummies(data[features], drop_first=True)
y = data[target]

print(f"Feature matrix shape: {X.shape}")
```

### 3. Visualize the Distribution of the Target Variable

```python
plt.figure(figsize=(8, 5))
plt.hist(y, bins=25, edgecolor="black", color="skyblue")
plt.xlabel(target)
plt.ylabel("Frequency")
plt.title(f"Distribution of {target}")
plt.tight_layout()
plt.show()
```

### 4. Train-Test Split & Model Training

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print(f"Training records: {len(X_train)}")
print(f"Testing records: {len(X_test)}")

# Train Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)
```

### 5. Evaluate the Model

```python
y_pred = model.predict(X_test)
results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})
print("\nACTUAL VS PREDICTED (first 10 rows)")
print(results.head(10).to_string(index=False))


mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("\nMODEL EVALUATION")
print(f"MAE: {mae:.4f}")
print(f"MSE: {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R2-score: {r2:.4f}")
```

### 6. Actual vs. Predicted Visualization

```python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, alpha=0.7, color="purple")
low = min(y_test.min(), y_pred.min())
high = max(y_test.max(), y_pred.max())
plt.plot([low, high], [low, high], "r--", label="Perfect prediction")
plt.xlabel(f"Actual {target}")
plt.ylabel(f"Predicted {target}")
plt.title(f"Actual vs Predicted {target}")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 7. Feature Coefficients

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
}).sort_values("Coefficient", key=abs, ascending=False)
print("\nFEATURE COEFFICIENTS")
print(coefficients.to_string(index=False))


results.to_csv("food_delivery_predictions.csv", index=False)
print("\nSaved: food_delivery_predictions.csv")
```
